```python
A digital locker requires users to enter a numerical passcode. As part of a security feature, the system checks the sum of the digits of the passcode.
 Write logic to compute the sum of the digits of a given number.
```

```python
password=int(input("enter the password"))
sum=0
while (password>0):
    digit=password%10
    sum=sum+digit
    password=password//10
print(sum)
```

```python
Scenario: In a secure login system, certain features are enabled only for users with prime-numbered user IDs.
 Write logic to check if a given number is prime.
```

```python
num=int(input("enter the value"))
if(num<2):
    print("neither prime nor composite")
else:
    count=0
    for i in range(1,num+1):
        if(num%i==0):
            count=count+1
    if(count>2):
        print("composite")
    else:
        print("prime")
        
    
    
```

```python
Scenario: A scientist is working on permutations and needs to calculate the factorial of numbers frequently.
 Write logic to find the factorial of a given number using recursion.
    
```

```python
def factorial(num):
    if(num==0) or (num==1):
        return 1
    else:
        return num*factorial(num-1)

num=int(input("enter the number"))

fact=factorial(num)
print(fact)


        
        
```
